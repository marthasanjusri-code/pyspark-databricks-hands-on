### WithColumn
-> Create new column

-> Can modify the existing column

-> use lit() for adding constant value

col("status") : References the values stored in the existing column named status

lit("status") : Creates a constant string column containing the text "status"

lit = The lit() function (short for literal) in PySpark converts a static Python value—such as a string, number, boolean, or date—into a PySpark Column object.


```python
# create new column naamed flag and put a constant string value "new" in all rows

df = df.withColumn('flag', lit("new"))
df.show()
```

### When-Otherwise

It is similar to case when statment in SQL

To create conditional columns


```python
# In Item_Type column, if meat is present then Non Veg else Veg

df_veg = df.withColumn(
    'Veg_flag',
    when(col('Item_Type')=='Meat', 'Non-Veg').otherwise('Veg')
)

df_veg.show(5)
```

```python
# create a flag expensive
# If Veg_flag : veg & Item_MRP < 100 then : Cheap else: costly
# otherwise : Non veg

df_veg.withColumn(
    'Expensive',
    when(((col('Veg_flag') == 'Veg') & (col('Item_MRP') < 100)), 'cheap') \
    .when(((col('Veg_flag') == 'Veg') & (col('Item_MRP') >= 100)), 'costly') \
    .otherwise('Non-Veg')
).show(5)

```

```python
# Right JOIN : we Don't want to delete any record from right dataframe 
df1.join(df2, df1['dept_id']==df2['dept_id'], 'right').show()

```

### Data Writing and Table

We load the raw data into temporary staging area in unity catalog volume (Data that need to be processed)

Managed tables : Data bricks handles both metadata and storage location.

stores at : structured delta lake tables

External tables : Data bricks handles only metadata and storage location handled by user (since they have own custom cloud path).

stores at : structured delta lake tables

---------------------------------------------
In general we create schema for bronze, silver, gold layer separatley and store transformed data table in them

So, lets create clean_data schema and store cleaned data table in it 

Data can be written in many formats like (delta lake, parquet, csv, json, avro)

Always default to Delta Lake for managed tables and intermediate storage inside Databricks.

---------------------
Data Writing Modes

Append : When you don't want to loose the data

overwrite : It overwrites the data which already exists

error : when you want to through error if file already exists

ignore : when you don't want to overwrite or through error or do anything , just ignore




```python
# Created new schema clean_data
# Write cleaned DataFrame to the clean_data schema
# after running this go to catalog and check
df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("pyspark_tutorial.clean_data.bigmart_sales")
```

Write cleaned data to csv format 

Databricks Volume is a secure, governed directory used to store and manage non-tabular (file-based) data.

so create,  (clean_data schema) -> (export_csv volume)

The df.coalesce(n) DataFrame Method (Partition Control)

What it does: Reduces the number of underlying Spark cluster partitions down to n without triggering a full data shuffle.

```python
csv_path = "/Volumes/pyspark_tutorial/clean_data/export_csv/bigmart_sales_csv"

# Write as a single CSV file with headers
df.coalesce(1).write \
    .mode("overwrite") \
    .option("header", "true") \
    .option("delimiter", ",") \
    .csv(csv_path)
```

#### Row based Format (like Avro)

```python
parquet_path = "/Volumes/pyspark_tutorial/clean_data/export_parquet/bigmart_sales_parquet"

# Write as Parquet
df.write \
    .mode("overwrite") \
    .parquet(parquet_path)
```
